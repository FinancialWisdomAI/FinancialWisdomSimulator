```python
"""
Figure 8.1
Stationary vs. Nonstationary Time Series

Lecture 8
Section 1, Concept 2

This program creates two time series:

1. A stationary series with a constant mean and constant variance.
2. A nonstationary series with an upward deterministic trend.
"""

import numpy as np
import matplotlib.pyplot as plt


# ==============================================================
# Figure settings
# ==============================================================

plt.rcParams.update({
    "font.size": 11,
    "axes.titlesize": 13,
    "axes.labelsize": 11,
    "xtick.labelsize": 10,
    "ytick.labelsize": 10,
    "lines.linewidth": 2.0,
    "figure.dpi": 150
})


# ==============================================================
# Generate the time series
# ==============================================================

# Reproducible random numbers
rng = np.random.default_rng(seed=42)

# Time periods
t = np.arange(1, 41)

# Random disturbances
epsilon_stationary = rng.normal(loc=0, scale=1.0, size=len(t))
epsilon_nonstationary = rng.normal(loc=0, scale=1.0, size=len(t))

# Stationary model:
# X_t = 10 + epsilon_t
stationary_mean = 10
stationary = stationary_mean + epsilon_stationary

# Nonstationary model:
# X_t = 5 + 0.25t + epsilon_t
trend = 5 + 0.25 * t
nonstationary = trend + epsilon_nonstationary


# ==============================================================
# Draw the figure
# ==============================================================

fig, axes = plt.subplots(
    nrows=1,
    ncols=2,
    figsize=(10, 4.4),
    sharey=True
)

# --------------------------------------------------------------
# Left panel: Stationary series
# --------------------------------------------------------------

axes[0].plot(
    t,
    stationary,
    color="navy",
    marker="o",
    markersize=3
)

axes[0].axhline(
    y=stationary_mean,
    color="firebrick",
    linestyle="--",
    linewidth=1.8
)

axes[0].set_title("Stationary Time Series")
axes[0].set_xlabel("Time")
axes[0].set_ylabel("Value")
axes[0].set_xlim(1, 40)
axes[0].set_ylim(3, 17)
axes[0].grid(True, alpha=0.25)

axes[0].text(
    15,
    stationary_mean + 1.5,
    "Constant mean",
    color="firebrick"
)


# --------------------------------------------------------------
# Right panel: Nonstationary series
# --------------------------------------------------------------

axes[1].plot(
    t,
    nonstationary,
    color="navy",
    marker="o",
    markersize=3
)

axes[1].plot(
    t,
    trend,
    color="firebrick",
    linestyle="--",
    linewidth=1.8
)

axes[1].set_title("Nonstationary Time Series")
axes[1].set_xlabel("Time")
axes[1].set_xlim(1, 40)
axes[1].grid(True, alpha=0.25)

axes[1].text(
    15,
    trend[23] - 0.3,
    "Changing mean",
    color="firebrick",
    rotation=28
)


# ==============================================================
# Finish and save
# ==============================================================

fig.tight_layout()

fig.savefig(
    "figure_8_1_stationary_vs_nonstationary.pdf",
    bbox_inches="tight"
)

fig.savefig(
    "figure_8_1_stationary_vs_nonstationary.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```
